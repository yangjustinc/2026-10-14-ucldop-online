# Lesson 18 · Programming Style

**Live coding | Software Carpentry**

How can we make Python code easier to read and maintain?

## Readable code

### 1 · Improve readability

Compare a short calculation with unclear names against one with meaningful names and consistent spacing.

## Assertions and documentation

### 2 · Check an assumption

Write a bulk-density function with an assertion that volume must be positive.

### 3 · Try an invalid input

Run the function with an invalid volume. Read the error.

### 4 · Give a function a docstring

Add documentation to an average function, then inspect it using `help()`.

## Challenges

### Challenge 18.1 · What Will Be Shown?

Predict which text will appear when someone requests help for a function, and which comments will not.

### Challenge 18.2 · Document This

Convert comments in a function that finds the middle of three values into a docstring. Inspect its help.

### Challenge 18.3 · Clean Up This Code

Refactor a difficult-to-read string-processing program without changing its output. (Longer exercise)
