# Lesson 16 · Writing Functions

**Live coding | Software Carpentry**

How do we turn repeated calculations into reusable functions?

## Define and call a function

### 1 · Create a greeting function

Define a function that prints three greeting lines.

### 2 · Run the function

Call the function you just created.

## Parameters and arguments

### 3 · Format a date

Define a function that accepts a year, month and day. Call it with positional arguments.

### 4 · Name the arguments

Call the date function again with keyword arguments in a different order.

## Return values

### 5 · Return an average

Create an average function that handles an empty list. Test both cases.

### 6 · Printed output versus returned value

Assign the result of the date-printing function to a variable. Inspect that variable.

## Challenges

### Challenge 16.1 · Identifying Syntax Errors

Identify and repair errors in a function definition. Read each error message before correcting it.

### Challenge 16.2 · Definition and Use

Predict what appears when a function name is printed without being called. Then make the function run.

### Challenge 16.3 · Order of Operations

Investigate calling a function before its definition. After fixing the order, inspect its returned value.

### Challenge 16.4 · Encapsulation

Write a function that loads a CSV file given its name and returns minimum values.

### Challenge 16.5 · Find the First

Return the first negative value from a list. Test a list with no negative values and an empty list.

### Challenge 16.6 · Calling by Name

Predict what happens when named arguments are supplied in a different order. Where have you used this before?

### Challenge 16.7 · Encapsulation of an If/Print Block

Turn an egg-mass classification into a function that **returns** the label. Then add warnings for unusual masses.

### Challenge 16.8 · Encapsulating Data Analysis

**Part A:** Load Japan’s GDP-per-capita data and calculate the average of the observations from the 1980s. (Longer exercise)

### Challenge 16.8 · Part B

Write a reusable function for a country, continent and year.

### Challenge 16.8 · Part C

Generalise the function so it discovers the available years in the requested decade.

### Challenge 16.9 · Simulating a Dynamical System

**Part A:** Create a function for the logistic population update. (Longer exercise)

### Challenge 16.9 · Part B

Start at 0.5, simulate ten updates, and collect all eleven values.

### Challenge 16.9 · Part C

Wrap the simulation in a function and compare 100 versus 1000 updates.

## Optional · Functions with Pandas

### 7 · Apply a rule to a column

Categorise life-expectancy values using a custom function and apply it to a Pandas column.
