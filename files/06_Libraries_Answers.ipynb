Rearrange the pieces into a working program that prints a random DNA base **and its position**. Fill the blanks; one import may be unnecessary.

### Challenge 6.5 · Importing with aliases

Convert π/2 radians into degrees using an alias for the mathematical module. Then try the same calculation without an alias. Which reads more clearly?

```python
import math as m
angle = m.degrees(m.pi / 2)
print(angle)
```

```python
import math
angle = math.degrees(math.pi / 2)
print(angle)
```

### Challenge 6.6 · Match the imports

Match each style of expression with a suitable import: `sin(pi/2)`, `m.sin(m.pi/2)` and `math.sin(math.pi/2)`. Is there an import style we should generally avoid?

```python
from math import sin, pi
print('sin(pi/2) =', sin(pi / 2))
```

```python
import math as m
print('sin(pi/2) =', m.sin(m.pi / 2))
```

```python
import math
print('sin(pi/2) =', math.sin(math.pi / 2))
```
