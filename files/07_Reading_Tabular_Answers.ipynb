## Live examples

### 1. Read the Oceania dataset

Open `data/gapminder_gdp_oceania.csv` and display the table.

```python
import pandas as pd
```

```python
data_oceania = pd.read_csv('data/gapminder_gdp_oceania.csv')
```

```python
data_oceania.head()
```

```python
data_oceania.tail()
```

### 2. Use country names as row labels

Read the file again so rows are identified by country rather than numbers.

```python
data_oceania_country = pd.read_csv('data/gapminder_gdp_oceania.csv', index_col='country')
print(data_oceania_country)
```

### 3. Inspect the table

Check its size, data types and available values.

```python
data_oceania_country.info()
```

### 4. See the column names

Display the labels used for the columns.

```python
data_oceania_country.columns
```

### 5. Transpose the table

Look at the same data with the rows and columns exchanged.

```python
data_oceania_country.T
```

### 6. Summarise numerical values

Generate descriptive statistics for the table.

```python
data_oceania_country.describe()
```
