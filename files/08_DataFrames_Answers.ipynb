## Select values

### 1. Choose a value by position

Load the European GDP data. Find the value in the first row and first column.

```python
import pandas as pd
```

```python
data = pd.read_csv('data/gapminder_gdp_europe.csv', index_col='country')
```

```python
data.head(1)
```

```python
data.iloc[0, 0]
```

### 2. Choose a value by name

Find the GDP-per-capita value for Albania in 1952 using labels.

```python
print(data.loc['Albania', 'gdpPercap_1952'])
```

### 3. Choose a whole row

Display all the years for Albania.

```python
print(data.loc['Albania', :])
```

### 4. Choose a whole column

Display the 1952 values for every country.

```python
print(data.loc[:, 'gdpPercap_1952'])
```

### 5. Select a rectangular region

Select countries from Italy through Poland and years from 1962 through 1972.

```python
subdata=data.loc['Italy':'Poland','gdpPercap_1962':'gdpPercap_1972']
```

### 6. Find a maximum

Find the largest value in each year of the selected region.

```python
subdata.max()
```

```python
subdata.max(axis=1)
```

### 7. Find a minimum

Find the smallest value in each year of the same region.

```python
subdata.min()
```

## Filter values

### 8. Test a condition

For every value in the selected region, ask whether it is greater than 10,000.

```python
subdata>10000
```

### 9. Use the test as a filter

Show values where the condition is met.

```python
subdata[subdata>10000]
```

### 10. Summarise the filtered values

Generate descriptive statistics using only values that meet the condition.

```python
subdata[subdata>10000].describe()
```

## Group values

### 11. Create a country score

For each European country, find the fraction of years in which its GDP per capita was above that year’s European average.

```python
mask_higher = data > data.mean()
```

```python
score = mask_higher.aggregate('sum', axis=1)/len(data.columns)
```

### 12. Group by the score

Group countries with the same score and summarise their GDP-per-capita data.

```python
data.groupby(score).median()
```

### Challenge 8.6 · Many Ways of Access

Select a column, a row, one value, several columns, several rows and a rectangular region. How can you preserve a two-dimensional table when selecting just one row or column?

```python
# One column: Series versus DataFrame
data['gdpPercap_2007']
data[['gdpPercap_2007']]

# One row: Series versus DataFrame
data.loc['Denmark']
data.loc[['Denmark']]

# One value
data.loc['Serbia', 'gdpPercap_2007']

# Several columns
data[['gdpPercap_1952', 'gdpPercap_2007']]

# Several rows
data.loc[['Denmark', 'Sweden']]

# Selected rows and columns
data.loc[['Denmark', 'Sweden'], ['gdpPercap_1952', 'gdpPercap_2007']]

# Named row and column ranges
data.loc['Denmark':'France', 'gdpPercap_1952':'gdpPercap_1962']
```
